#### Exercício 1

```python
usd_texto = input("Digite a quantidade de Dólares (USD) que você possui: ")
cotacao_texto = input("Digite a cotação atual do Dólar em Reais (BRL): ")

# convertendo as strings do input para float
usd = float(usd_texto)
cotacao = float(cotacao_texto)

# calculando em reais
brl = usd * cotacao

# exibindo o resultado formatado com 2 casas decimais
print(f"Seus USD {usd:.2f} equivalem a R$ {brl:.2f} na cotação atual.")
```

#### Exercício 2

```python
valor_inicial = float(input("Digite o Valor Inicial do Investimento (R$): "))
valor_final = float(input("Digite o Valor Final do Investimento (R$): "))

# calculando o ROI
roi = ((valor_final - valor_inicial) / valor_inicial) * 100

print(f"O Retorno Sobre Investimento (ROI) foi de {roi:.2f}%")
```

#### Exercício 3

```python
ativo1 = "   petr4 "
ativo2 = "  iTuB4  "
ativo3 = "   VALE3  "
```

```python
# limpando os espaços (strip) e padronizando para maiúsculas (upper)
ativo1_limpo = ativo1.strip().upper()
ativo2_limpo = ativo2.strip().upper()
ativo3_limpo = ativo3.strip().upper()

# exibindo os ativos formatados com f-strings
print(f"Ativo da carteira: {ativo1_limpo}")
print(f"Ativo da carteira: {ativo2_limpo}")
print(f"Ativo da carteira: {ativo3_limpo}")
print("-" * 30)
```

#### Exercício 4

```python
try:
    total_dividendos = float(input("Digite o valor total dos dividendos (R$): "))
    numero_cotistas = int(input("Digite o número de cotistas: "))
    
    valor_por_cotista = total_dividendos / numero_cotistas
    
    print(f"Cada cotista receberá R$ {valor_por_cotista:.2f}")
    
except ValueError:
    print("Erro: Por favor, digite apenas números válidos (use ponto para separar os centavos).")
except ZeroDivisionError:
    print("Erro: O número de cotistas não pode ser zero! Não é possível dividir por zero.")
```
